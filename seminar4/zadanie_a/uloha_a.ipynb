```python

# uloha 1 

import pandas as pd
import pyarrow.parquet as pq
pf = pq.ParquetFile("/Users/ixsaiw/ukf/DatabazoveTechnologie/seminar4/zadanie_a/data/green_tripdata_2024-01.parquet")
md = pf.metadata
print(md.num_rows)
print(md.num_columns)
print(md.num_row_groups)
print(pf.schema_arrow) 
print(md.created_by)

```

```python

# Uloha 2

for i in range(md.num_row_groups):
    rg = md.row_group(i)
    print(
        f"Row group {i}: "
        f"riadkov={rg.num_rows}, "
        f"stĺpcov={rg.num_columns}, "
        f"veľkosť={rg.total_byte_size / (1024**2):.2f} MB"
    )
```

```python

# uloha 3

print(pf.schema_arrow)
col_name =  "passenger_count"
rg = md.row_group(0)
idx = next(j for j in range(rg.num_columns) if rg.column(j).path_in_schema == col_name)
col = rg.column(idx)

print("Názov stĺpca:        ", col.path_in_schema)
print("Fyzický dátový typ:  ", col.physical_type)
print("Kompresia:           ", col.compression)
print("Počet hodnôt:        ", col.num_values)
print("Komprimovaná veľkosť:", col.total_compressed_size, "B")
print("Nekomprimovaná veľk.:", col.total_uncompressed_size, "B")

stats = col.statistics
if stats is not None and stats.has_min_max:
    print("Minimum:             ", stats.min)
    print("Maximum:             ", stats.max)
else:
    print("Min/max štatistiky nie sú v súbore uložené")

if stats is not None:
    print("Počet NULL hodnôt:   ", stats.null_count)

```
